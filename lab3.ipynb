```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("StudentsPerformance.csv")
print(df.head())
```

```python
arr = df.to_numpy()
```

```python
data1=arr[ :1000 , 6 ].tolist()
data2=arr[ :1000 , 7 ].tolist()
print(data1,data2)
```

```python
plt.scatter(data1,data2)
plt.xlabel('reading score')
plt.ylabel('writing score')
plt.title('Scatter plot')
plt.show()
```

```python
mean1 = np.mean(data1)
mean2 = np.mean(data2)
n = len(data1)
print(mean1, mean2)
```

```python
# cov_manual=np.sum()
cov_manual=np.sum((data1 - mean1)*(data2-mean2))/n
print (cov_manual)
```

```python
cov_M=np.cov(data1,data2, bias=True)
print(cov_M)
```

```python
cov = cov_M[0,1]
print(cov)
```

```python
var1 = cov_M[0,0]
var2 = cov_M[1,1]
print (var1, var2)
```

```python
r=cov/(np.sqrt(var1*var2))
r
```

```python
alpha = 0.05
```

```python
TS = (r*(np.sqrt(n-2)))/(np.sqrt(1-r**2))
print(TS)
```

```python
crit_value = stats.t.ppf(1-alpha,n-2)
print(crit_value)
```
