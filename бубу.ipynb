```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df=pd.read_csv ('screen_time_behavior.csv')
print(df.head())
```

```python
arr = df.to_numpy()
data=arr[:,6]
print(data)
```

```python
mean=np.mean(data)
print(mean)
```

```python
sigma = 1.5
alpha = 0.05
```

```python
z=stats.norm.ppf(0.975, loc=0, scale=1)
print(z)
```

```python
n=len(data)
print(n)
```

```python
import math 
```

```python
margin=z*sigma/math.sqrt(n)
print(margin)
```

```python
lower_boundary=mean-margin
upper_boundary=mean+margin
print("Confidence interval(",lower_boundary,upper_boundary,")")
```

```python
from scipy.stats import norm
confidence = 1-alpha
lower,upper=stats.norm.interval(confidence,loc=mean,scale=sigma/np.sqrt(n))
print(lower,upper)
```

```python
v=np.var(data)
print(v)
```

```python
t=stats.t.ppf(0.975,n-1)
print(t)
```

```python
s=math.sqrt(np.var(data,ddof=1))
print(s)
```

```python
margin=t*s/math.sqrt(n)
print(margin)
```

```python
lower_boundary=mean-margin
upper_boundary=mean+margin
print("Confidence interval(",lower_boundary,upper_boundary,")")
```

```python
dof=n-1
sem=np.std(data,ddof=1)/np.sqrt(n)
confidence=1-alpha
interval=stats.t.interval(confidence,dof,loc=mean,scale=sem)
print(interval)
```
